## Предельные циклы и отображение Пуанкаре
Рассмотрим систему дифференциальных уравнений:
$$
\begin{cases}
\displaystyle \frac{dx}{dt}=-ky+mx\sin(x^2+y^2),\\[6pt]
\displaystyle \frac{dy}{dt}=kx+my\sin(x^2+y^2).
\end{cases}
$$
Докажите, что у неё есть предельные циклы. Для предельного цикла с номером \(m+1\), считая от начала координат, постройте сечение Пуанкаре и численно — отображение Пуанкаре.

Какое собственное число имеет линеаризованное отображение Пуанкаре? Как связано это число с устойчивостью или неустойчивостью выбранного предельного цикла? Какие выводы можно сделать по этому числу?

## Решение

**Сведём исследование расстояния к одному уравнению**, имеем:
При наших постоянные, где $m=2$, $k=17$ дана система
$$
\begin{cases}
x'=-17y+2x\sin(x^2+y^2),\\
y'=17x+2y\sin(x^2+y^2).
\end{cases}
$$

Как в примере с осциллятором, умножаем первое уравнение на $x$, второе — на $y$:
$$
xx'=-17xy+2x^2\sin(x^2+y^2),
$$
$$
yy'=17xy+2y^2\sin(x^2+y^2).
$$

Складываем. Слагаемые $-17xy$ и $17xy$ сокращаются:
$$
xx'+yy'=2(x^2+y^2)\sin(x^2+y^2).
$$

Поскольку $(x^2+y^2)'=2xx'+2yy',$ получаем $(x^2+y^2)'=4(x^2+y^2)\sin(x^2+y^2).$ Обозначим $u=x^2+y^2.$
Число $u$ — квадрат расстояния до начала координат. Для него получили одно скалярное уравнение:
$$
u'=4u\sin u,\qquad u\geqslant0.
$$

**Найдём периодические траектории**. Сначала найдём постоянные решения полученного уравнения: $4u\sin u=0.$

Следовательно,
$$
u=0
\quad\text{или}\quad
u=n\pi,\qquad n=1,2,\ldots
$$

Значению $u=0$ соответствует стационарная точка $(0,0)$.

При $u=n\pi>0$:
$$
x^2+y^2=n\pi,\qquad \sin(x^2+y^2)=0.
$$

Поэтому на такой окружности исходная система принимает вид
$$
\begin{cases}
x'=-17y,\\
y'=17x.
\end{cases}
$$

Дифференцируем первое уравнение и подставляем второе:
$$
x''=-17y'=-17\cdot17x=-289x.
$$

Получили уравнение гармонического осциллятора: $x''+289x=0.$
Для начальной точки $(r_n,0)$, где $r_n=\sqrt{n\pi}$, решение имеет вид
$$
x(t)=r_n\cos(17t),\qquad
y(t)=r_n\sin(17t).
$$

Действительно, ведь: $x^2(t)+y^2(t)=r_n^2\bigl(\cos^2(17t)+\sin^2(17t)\bigr)=n\pi.$

Таким образом, траектории — окружности
$$
x^2+y^2=n\pi,\qquad n=1,2,\ldots
$$

Первое возвращение в исходную точку происходит при $17T=2\pi$. Поэтому
$$
T=\frac{2\pi}{17}\approx0.369599136.
$$

В точке $(r_n,0)$ имеем $y'=17r_n>0$, поэтому движение направлено против часовой стрелки.

Доказываем изолированность и определяем устойчивость. Между соседними значениями $n\pi$ и $(n+1)\pi$ функция $\sin u$ сохраняет ненулевой знак. Поэтому $u'=4u\sin u$, также сохраняет знак: $u(t)$ строго увеличивается либо строго уменьшается.

Следовательно, такое решение не может вернуться к исходному значению $u$, а значит, соответствующая траектория не может быть периодической. Стационарные значения $u=n\pi$ решения не пересекают вследствие единственности решения гладкого скалярного ОДУ.

Таким образом, каждая найденная окружность — изолированная периодическая траектория, то есть предельный цикл.

Для устойчивости исследуем направление изменения $u$ около $u=n\pi$:
При нечётном $n$ слева от $n\pi$ получается $u'>0$ и справа от $u'<0$. Поведение: Решения приближаются к циклу.
При чётном $n$ слева от $n\pi$ получается $u'<0$ и справа от $u'>0$. Поведение: Решения удаляется к циклу.

Следовательно, нечётные циклы притягивают, чётные — отталкивают.

Нужный номер цикла: $n=m+1=3.$
Поэтому
$$
x^2+y^2=3\pi,\qquad R=\sqrt{3\pi}\approx3.069980124.
$$

В частности,
$$
2\pi<u<3\pi\ \Longrightarrow\ u'>0,
$$
$$
3\pi<u<4\pi\ \Longrightarrow\ u'<0.
$$

С обеих сторон решения приближаются к $u=3\pi$, следовательно, расстояние до выбранной окружности стремится к нулю.

**Строим отображение Пуанкаре**. Выбираем небольшой горизонтальный отрезок $\Sigma=\{(x,0):\ |x-R|<0.15\}.$
На нём $x>0$ и $y'=17x>0.$ Поэтому траектории пересекают отрезок поперечно.

На сечении удобно задавать точку числом $u_0=x_0^2.$

Поскольку $x_0>0$, по этому числу однозначно восстанавливается точка:
$$
(x_0,y_0)=(\sqrt{u_0},0).
$$

Проверим время возвращения. Если $u(t)$ решает
$$
u'=4u\sin u,\qquad u(0)=u_0,
$$
то непосредственной подстановкой проверяется решение исходной системы:
$$
x(t)=\sqrt{u(t)}\cos(17t),\qquad
y(t)=\sqrt{u(t)}\sin(17t).
$$
Здесь
$$
\bigl(\sqrt u\bigr)'=\frac{u'}{2\sqrt u}
=2\sqrt u\sin u.
$$
Поэтому
$$
x'=2\sqrt u\sin u\cos(17t)-17\sqrt u\sin(17t)
=2x\sin u-17y,
$$
$$
y'=2\sqrt u\sin u\sin(17t)+17\sqrt u\cos(17t)
=2y\sin u+17x.
$$

Первое возвращение на положительную полуось происходит через
$$
T=\frac{2\pi}{17}.
$$
Отображение Пуанкаре сопоставляет начальному значению $u_0$ значение после одного оборота:
$$
P(u_0)=u(T;u_0).
$$
Для выбранного цикла:
$$
P(3\pi)=3\pi.
$$
Следовательно, $u_*=3\pi$ — неподвижная точка отображения.

**Находим собственное число отображения**. Обозначим правую часть скалярного уравнения:
$F(u)=4u\sin u.$ Её производная: $F'(u)=4\sin u+4u\cos u.$

В точке выбранного цикла:
$$
F'(3\pi)
=4\sin(3\pi)+12\pi\cos(3\pi)
=0-12\pi=-12\pi.
$$
Чтобы определить, как малое изменение начального значения влияет на решение, рассмотрим производную решения по начальному значению:
$$
q(t)=\left.\frac{\partial u(t;u_0)}{\partial u_0}\right|_{u_0=3\pi}
$$
удовлетворяет задаче $q'=-12\pi q, q(0)=1.$
Её решение: $q(t)=e^{-12\pi t}.$
Поэтому собственное число одномерного линеаризованного отображения Пуанкаре равно
$$
\mu=P'(3\pi)=q(T)=e^{-12\pi T}.
$$
Подставляем период:
$$
\mu=\exp\left(-\frac{24\pi^2}{17}\right)
\approx8.88652866\cdot10^{-7}.
$$
Так как
$$
|\mu|<1,
$$
отображение вблизи неподвижной точки является сжатием. После каждого оборота малые отклонения на сечении уменьшаются. Следовательно, цикл орбитально асимптотически устойчив.

**Выполняем численную проверку**. Величина $q(t)$ показывает, во сколько раз изменяется малое начальное отклонение в линейном приближении:
$$
\text{отклонение в момент }t
\approx q(t)\cdot\text{начальное отклонение}.
$$
В начальный момент отклонение ещё не изменилось, поэтому $q(0)=1$.

Численно решаем задачу
$$
\boxed{q'=-12\pi q,\qquad q(0)=1}
$$
до момента
$$
T=\frac{2\pi}{17},
$$
то есть в течение одного полного оборота.

Искомое собственное число — значение в конце этого промежутка:
$$
\mu=q(T).
$$

При помощи метода Рунге – Кутты четвёртого порядка получаем, что:

При 400 шагов за оборот = $8.886530225\cdot10^{-7}$

При 800 шагов за оборот = $8.886528758\cdot10^{-7}$

При 1600 шагов за оборот = $8.886528667\cdot10^{-7}$

```python
import numpy as np
import matplotlib.pyplot as plt

%matplotlib inline
%config InlineBackend.figure_format = 'retina'

plt.close('all')

# Постоянные
m = 2
k = 17

# Палитра
brown       = [0.4627, 0.3294, 0.2314]  # коричневый
light_brown = [0.7451, 0.5843, 0.4353]  # светло-коричневый
beige       = [0.8549, 0.7765, 0.6667]  # бежевый
gray        = [0.8000, 0.8000, 0.8000]  # серая сетка
background  = [1.0000, 1.0000, 1.0000]  # белый фон

plt.rcParams.update({
    'figure.facecolor': background, 'axes.facecolor': background,
    'savefig.facecolor': background, 'grid.color': gray,
    'grid.alpha': 1, 'axes.grid': True, 'font.size': 11,
    'axes.prop_cycle': plt.cycler(color=[brown, light_brown, gray]),
    'figure.figsize': (10, 4), 'figure.dpi': 120,
})
```

```python
# Метод Рунге — Кутты 4-го порядка
def rk4(rhs, t, z0):
    t = np.asarray(t, dtype=float)
    z0 = np.asarray(z0, dtype=float)

    z = np.zeros((len(t),) + z0.shape)
    z[0] = z0

    for j in range(len(t) - 1):
        h = t[j + 1] - t[j]
        tj = t[j]
        zj = z[j]

        k1 = rhs(tj, zj)
        k2 = rhs(tj + h / 2, zj + h * k1 / 2)
        k3 = rhs(tj + h / 2, zj + h * k2 / 2)
        k4 = rhs(tj + h, zj + h * k3)

        z[j + 1] = zj + h * (k1 + 2 * k2 + 2 * k3 + k4) / 6

    return z
```

```python
# Параметры выбранного цикла
n = m + 1
R = np.sqrt(n * np.pi)
T = 2 * np.pi / k

linear_rate = 2 * m * n * np.pi * (-1)**n
mu_exact = np.exp(linear_rate * T)

# Радиальное уравнение
def radial_rhs(t, delta):
    # Здесь delta = r - R — отклонение радиуса от цикла.
    angle = 2 * R * delta + delta**2

    return m * (-1)**n * (R + delta) * np.sin(angle)


# Отображение Пуанкаре
def poincare(delta0, steps=1600):
    t_return = np.linspace(0, T, steps + 1)

    # Отклонение после одного оборота: P(R + delta0) - R.
    return rk4(radial_rhs, t_return, delta0)[-1]


delta0 = np.linspace(-0.15, 0.15, 121)
p_delta = poincare(delta0)

# Уравнение в вариациях: q' = linear_rate * q, q(0) = 1.
t_period = np.linspace(0, T, 1601)

mu_num = rk4(
    lambda t, q: linear_rate * q,
    t_period,
    1.0
)[-1]

print(f'Радиус цикла:       R = {R:.12f}')
print(f'Период:            T = {T:.12f}')
print(f'Собственное число: μ = {mu_exact:.12e}')
print(f'Численная проверка:   {mu_num:.12e}')

# Проверка производной центральной разностью
print('\nПроверка центральной разностью:')

for eps0 in [1e-2, 1e-3, 1e-4]:
    pp = poincare(np.array([-eps0, eps0]))
    derivative = (pp[1] - pp[0]) / (2 * eps0)

    print(f'ε = {eps0:.0e}:  μ ≈ {derivative:.12e}')

# Проверка при уменьшении шага вдвое
refined = poincare(delta0, steps=3200)
error = np.max(np.abs(refined - p_delta))

print(f'\nИзменение при удвоении числа шагов: {error:.3e}')

assert abs(mu_num / mu_exact - 1) < 1e-8
assert error < 1e-12
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15.5, 5.5))
fig.set_facecolor(background)

# Оформление осей
for ax in axes:
    ax.set_facecolor(background)
    ax.set_axisbelow(True)

    ax.grid(True, color=gray, linewidth=0.7)

    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.spines['left'].set_color(gray)
    ax.spines['bottom'].set_color(gray)

    ax.tick_params(labelsize=10, length=0, pad=7)

initial_conditions = [
    (-0.15, light_brown, '--'),
    (0.15, brown, '-')
]


# Первый график: циклы, близкие траектории и сечение
phi = np.linspace(0, 2 * np.pi, 1001)

for nn in [n - 1, n, n + 1]:
    radius = np.sqrt(nn * np.pi)
    selected = nn == n

    axes[0].plot(
        radius * np.cos(phi),
        radius * np.sin(phi),
        color=brown if selected else beige,
        linestyle='-' if selected else ('--' if nn < n else ':'),
        linewidth=2.3 if selected else 1.5,
        label=f'Цикл {nn}'
    )

t_two_periods = np.linspace(0, 2 * T, 3201)

for d0, color, style in initial_conditions:
    delta = rk4(radial_rhs, t_two_periods, d0)
    radius = R + delta

    axes[0].plot(
        radius * np.cos(k * t_two_periods),
        radius * np.sin(k * t_two_periods),
        color=color,
        linestyle=style,
        linewidth=1.2,
        label=rf'$\delta_0={d0:+.2f}$'
    )

axes[0].plot(
    [R - 0.15, R + 0.15],
    [0, 0],
    color=brown,
    linewidth=4,
    solid_capstyle='round',
    label='Сечение'
)

axes[0].set_aspect('equal', adjustable='box')
axes[0].set(
    xlim=(-4, 4),
    ylim=(-4, 4),
    xlabel='$x$',
    ylabel='$y$'
)
axes[0].set_title('Третий цикл и сечение', fontsize=13, pad=18)


# Второй график: отображение Пуанкаре
axes[1].plot(
    delta0, p_delta * 1e6,
    color=brown,
    linewidth=2.2,
    label='Численное отображение'
)

axes[1].plot(
    delta0, mu_exact * delta0 * 1e6,
    color=light_brown,
    linestyle='--',
    linewidth=2,
    label=r'$\mu\delta_0$'
)

axes[1].set(
    xlim=(-0.15, 0.15),
    xlabel=r'$\delta_0=r_0-R$',
    ylabel=r'$10^6(P(R+\delta_0)-R)$'
)
axes[1].set_xticks(
    [-0.15, -0.10, -0.05, 0, 0.05, 0.10, 0.15]
)
axes[1].set_title('Отображение Пуанкаре', fontsize=13, pad=18)


# Третий график: приближение к циклу
for d0, color, style in initial_conditions:
    delta = rk4(radial_rhs, t_period, d0)

    axes[2].semilogy(
        t_period, np.abs(delta),
        color=color,
        linestyle=style,
        linewidth=2,
        label=rf'$\delta(0)={d0:+.2f}$'
    )

axes[2].set(
    xlim=(0, T),
    xlabel='$t$',
    ylabel=r'$|r(t)-R|$'
)
axes[2].set_title('Сближение за один оборот', fontsize=13, pad=18)
axes[2].minorticks_off()


# Легенды под графиками
for j, ax in enumerate(axes):
    ax.legend(
        loc='upper center',
        bbox_to_anchor=(0.5, -0.23),
        ncol=2 if j == 0 else 1,
        facecolor=background,
        edgecolor=gray,
        framealpha=1,
        fontsize=9,
        handlelength=2.7,
        borderpad=0.8,
        labelspacing=0.6
    )

plt.tight_layout(pad=1.8, w_pad=2.7)
plt.show()
```

**Комментарий к графикам и ответ**. Слева видно, что близкие траектории приближаются к третьей окружности и изнутри, и снаружи. В центре показано отклонение от цикла после одного оборота. Оно значительно меньше начального. Для наглядности значения по вертикальной оси увеличены в миллион раз. Справа видно, как расстояние до выбранной окружности уменьшается со временем.

Система имеет бесконечно много предельных циклов:
$$
r_n=\sqrt{n\pi},\qquad n=1,2,\ldots
$$
Выбранный третий цикл:
$$
x^2+y^2=3\pi.
$$
Для него
$$
T=\frac{2\pi}{17},\qquad
\mu\approx8.89\cdot10^{-7}<1.
$$
Поэтому малые отклонения после каждого оборота уменьшаются. Третий цикл орбитально асимптотически устойчив: близкие траектории со временем приближаются к этой окружности.

К нему приближаются все траектории, начинающиеся между второй и четвёртой окружностями:
$$
\sqrt{2\pi}<r<\sqrt{4\pi}.
$$
Это кольцо является областью притяжения третьего цикла.
