## Маятник с трением и периодической внешней силой
**Задание 3.3.** Анализ математического маятника с трением и внешней силой

Рассмотрим физический маятник, состоящий из жёсткого стержня длиной $l$ и груза массой $m_0$ на его конце, на который действует ускорение свободного падения $g$. Маятник затухает из-за наличия силы трения, пропорциональной его угловой скорости, которую характеризует коэффициент затухания $b$. Кроме того, на маятник действует внешняя периодическая сила с синусоидальной формой $F\sin(\alpha t)$, где $F$ — амплитуда силы, а $\alpha$ — угловая частота внешней силы.

Уравнение движения маятника задаётся нелинейным дифференциальным уравнением второго порядка:

$$
\frac{d^2\theta}{dt^2}
+\frac{b}{m_0}\frac{d\theta}{dt}
+\frac{g}{l}\sin(\theta)
=F\sin(\alpha t),
$$

где $\theta$ представляет собой угловое смещение от вертикали.

**Параметры задачи.**

- Длина маятника $l$: 1 м.
- Масса груза $m_0$: 1 кг.
- Ускорение свободного падения $g$: 9.81 м/с².
- Коэффициент затухания $b$: k/50 кг·м²/с, $k$ — номер студента в списке.
- Амплитуда силы $F$: m/10 $Н$, $m$ — номер группы.
- Начальное угловое смещение $\theta_0$: $\pi/4$ рад.
- Начальная угловая скорость $\omega_0$: 0 рад/с.
- Время интегрирования: $0\le t\le100$ с.

3. **Численное решение.** Используя MATLAB, численно решите систему уравнений первого порядка для ряда значений $\alpha$, чтобы исследовать выход на периодический режим маятника. Ваша симуляция должна охватывать диапазон значений $\alpha$ от 0 до 5 рад/с с интервалами в 0.05 рад/с.

## Решение

Решить $$\theta''+\frac b{m_0}\theta'+\frac gl\sin\theta=F\sin(\alpha t),$$

при $l=1$, $m_0=1$, $g=9.81$, $b=k/50$, $F=m/10$, $\theta(0)=\pi/4$, $\theta'(0)=0$, $0\le t\le100$.

Частоты: $\alpha=0,0.05,\ldots,5$. Построить амплитудно-частотную характеристику после затухания переходного процесса, объяснить резонанс и влияние трения.

**Переходём к системе первого порядка**. Вводим угловую скорость $\omega=\theta'$. Тогда
$$\begin{cases}\theta'=\omega,\\\omega'=-\dfrac b{m_0}\omega-\dfrac gl\sin\theta+F\sin(\alpha t).\end{cases}$$

Начальные условия:
$$\theta(0)=\frac\pi4\approx0.7853981634,\qquad\omega(0)=0.$$

Используется именно уравнение, напечатанное в условии. Масса $m_0$ и номер группы $m$ — разные обозначения.

**Подставляем параметры**. $$b=\frac{17}{50}=0.34,\qquad F=\frac2{10}=0.2,$$
$$\theta'=\omega,\qquad\omega'=-0.34\omega-9.81\sin\theta+0.2\sin(\alpha t).$$

Частоты изменяем от 0 до 5 с шагом 0.05. Число расчётов:
$$N_\alpha=\frac{5-0}{0.05}+1=101.$$

**Находим линейную оценку амплитуды**. В установившемся режиме углы окажутся небольшими. При малых углах $\sin\theta\approx\theta$: $\theta''+0.34\theta'+9.81\theta=0.2\sin(\alpha t)$.

Ищем установившееся решение:
$$
\theta_s=u\sin(\alpha t)+v\cos(\alpha t).
$$

Его производные:
$$
\theta_s'=\alpha u\cos(\alpha t)-\alpha v\sin(\alpha t),
$$
$$
\theta_s''=-\alpha^2u\sin(\alpha t)-\alpha^2v\cos(\alpha t).
$$

Подставляем в уравнение и собираем слагаемые:
$$
\left((9.81-\alpha^2)u-0.34\alpha v\right)\sin(\alpha t)
+
\left(0.34\alpha u+(9.81-\alpha^2)v\right)\cos(\alpha t)
=
0.2\sin(\alpha t).
$$

Следовательно:
$$
\begin{cases}
(9.81-\alpha^2)u-0.34\alpha v=0.2,\\
0.34\alpha u+(9.81-\alpha^2)v=0.
\end{cases}
$$

Обозначим:
$$
\Delta=9.81-\alpha^2,\qquad c=0.34\alpha,
\qquad D=\Delta^2+c^2.
$$

Умножая первое уравнение на $\Delta$, второе — на $c$, и складывая, получаем:
$$
\Delta^2u-\Delta cv+c^2u+c\Delta v=0.2\Delta,
$$

$$
Du=0.2\Delta,
\qquad
u=\frac{0.2\Delta}{D}.
$$

Аналогично, умножая первое уравнение на $-c$, второе — на $\Delta$, получаем:
$$
Dv=-0.2c,
\qquad
v=-\frac{0.2c}{D}.
$$

Амплитуда:
$$
A_{\mathrm{lin}}=\sqrt{u^2+v^2}
=\sqrt{\frac{0.2^2\Delta^2+0.2^2c^2}{D^2}}
=\sqrt{\frac{0.2^2D}{D^2}}
=\frac{0.2}{\sqrt D}.
$$

Таким образом, при $\alpha>0$:
$$
A_{\mathrm{lin}}(\alpha)
=\frac{0.2}{\sqrt{(9.81-\alpha^2)^2+0.34^2\alpha^2}}.
$$

**Находим резонансную частоту линейной модели**. Максимум амплитуды соответствует минимуму:
$$
D(\alpha)=(9.81-\alpha^2)^2+0.1156\alpha^2.
$$

Дифференцируем:
$$
D'(\alpha)
=2(9.81-\alpha^2)(-2\alpha)+0.2312\alpha,
$$
$$
D'(\alpha)=2\alpha\bigl(-2(9.81-\alpha^2)+0.1156\bigr).
$$

При \(\alpha>0\) условие \(D'(\alpha)=0\) даёт:
$$
-19.62+2\alpha^2+0.1156=0,
$$
$$
2\alpha^2=19.5044,
\qquad
\alpha^2=9.7522.
$$

Следовательно:
$$
\alpha_{\mathrm{рез,lin}}=\sqrt{9.7522}
\approx3.12285\ \text{рад/с}.
$$

Здесь $D'$ меняет знак с минуса на плюс, поэтому найден минимум знаменателя.

Подставляем частоту в формулу амплитуды:
$$
A_{\max,\mathrm{lin}}
=
\frac{0.2}
{\sqrt{(9.81-9.7522)^2+0.1156\cdot9.7522}},
$$
$$
A_{\max,\mathrm{lin}}
=
\frac{0.2}{\sqrt{0.0578^2+1.12735432}}
=
\frac{0.2}{\sqrt{1.13069516}}
\approx0.18809\ \text{рад}.
$$

**Оцениваем время затухания переходного процесса**. Для однородного линейного уравнения:
$$
\theta''+0.34\theta'+9.81\theta=0
$$
характеристическое уравнение имеет вид:
$$
\lambda^2+0.34\lambda+9.81=0.
$$

Вычисляем дискриминант:
$$
D_\lambda=0.34^2-4\cdot9.81
=0.1156-39.24=-39.1244.
$$

Отсюда:
$$
\lambda_{1,2}
=\frac{-0.34\pm i\sqrt{39.1244}}{2}
\approx-0.17\pm3.12748i.
$$

Переходное решение:
$$
\theta_h(t)=e^{-0.17t}
\bigl(C_1\cos(3.12748t)+C_2\sin(3.12748t)\bigr).
$$

Его амплитуда убывает с множителем $e^{-0.17t}$. Используя начальное отклонение как ориентировочный масштаб, потребуем:
$$
\frac{\pi}{4}e^{-0.17t}<10^{-4}.
$$

Тогда:
$$
e^{-0.17t}<\frac{4\cdot10^{-4}}{\pi},
$$
$$
t>\frac{\ln\!\left(\dfrac{\pi}{4\cdot10^{-4}}\right)}{0.17}
\approx52.76\ \text{с}.
$$

Берём запас и анализируем участок $60\le t\le100$. При $t=60$:
$$
\frac{\pi}{4}e^{-0.17\cdot60}
=\frac{\pi}{4}e^{-10.2}
\approx2.92\cdot10^{-5}\ \text{рад}.
$$

Это предварительная оценка по линейной модели; для нелинейной системы проверяем результат численно.

**Вычисляем установившуюся амплитуду**. Амплитуда — половина размаха за полный период:
$$
A=\frac{\theta_{\max}-\theta_{\min}}{2}.
$$
Но при $\alpha=0.05$:
$$
T=\frac{2\pi}{0.05}=40\pi\approx125.66\ \text{с},
$$
что больше всего времени расчёта. Поэтому размах участка [60,100] нельзя принимать за полный размах.

По значениям решения на этом участке приближённо восстанавливаем периодическую функцию:
$$
\theta_s(t)\approx
a_1\sin(\alpha t)+b_1\cos(\alpha t)
+a_3\sin(3\alpha t)+b_3\cos(3\alpha t).
$$
Коэффициенты выбираем методом наименьших квадратов:
$$
\sum_i
\left(
a_1\sin(\alpha t_i)+b_1\cos(\alpha t_i)
+a_3\sin(3\alpha t_i)+b_3\cos(3\alpha t_i)
-\theta_i
\right)^2
\longrightarrow\min.
$$

Затем вводим фазу $\varphi=\alpha t$ и рассматриваем полный период:
$$
s(\varphi)=
a_1\sin\varphi+b_1\cos\varphi
+a_3\sin3\varphi+b_3\cos3\varphi,
\qquad 0\le\varphi\le2\pi.
$$

Искомая оценка:
$$
A(\alpha)\approx
\frac{\max\limits_{0\le\varphi\le2\pi}s(\varphi)
-\min\limits_{0\le\varphi\le2\pi}s(\varphi)}{2}.
$$

При $\alpha=0$:
$$
0.2\sin(0t)=0.
$$

Внешнее воздействие отсутствует, колебания затухают, поэтому установившаяся амплитуда:
$$
A(0)=0.
$$

**Проверяем точность и записываем результат**. В Python повторяем расчёт RK4 с шагами $h=0.01$ и $h=0.005$.

В выполненном Python-расчёте:
$$
\max|\theta_s-\theta|\approx3.2\cdot10^{-5}\ \text{рад},
$$
$$
\max_\alpha|A_{[60,100]}-A_{[70,100]}|
\approx3.4\cdot10^{-6}\ \text{рад},
$$
$$
\max_\alpha|A_h-A_{h/2}|
\approx1.1\cdot10^{-8}\ \text{рад}.
$$

Все три величины меньше выбранной точности $10^{-4}$ рад.

Максимум на заданной сетке частот:
$$
\alpha=3.10\ \text{рад/с},
\qquad A\approx0.187646585\ \text{рад}.
$$

Переводим в градусы:
$$
A_{\mathrm{град}}
=0.187646585\cdot\frac{180}{\pi}
\approx10.75^\circ.
$$

Пик около собственной частоты $\sqrt{9.81}\approx3.132$ рад/с соответствует резонансу. Трение ограничивает амплитуду: при увеличении затухания резонансный пик становится ниже и шире.

```python
import numpy as np
import matplotlib.pyplot as plt

%matplotlib inline
%config InlineBackend.figure_format = 'retina'

plt.close('all')

# Постоянные
m = 2
k = 17

# Палитра
brown       = [0.4627, 0.3294, 0.2314]  # коричневый
light_brown = [0.7451, 0.5843, 0.4353]  # светло-коричневый
beige       = [0.8549, 0.7765, 0.6667]  # бежевый
gray        = [0.8000, 0.8000, 0.8000]  # серая сетка
background  = [1.0000, 1.0000, 1.0000]  # белый фон

# Оформление графиков
plt.rcParams.update({
    'figure.facecolor': background,
    'axes.facecolor': background,
    'savefig.facecolor': background,
    'grid.color': gray,
    'grid.alpha': 1,
    'axes.grid': True,
    'font.size': 11,
    'axes.prop_cycle': plt.cycler(
        color=[brown, light_brown, gray]
    ),
    'figure.figsize': (10, 4),
    'figure.dpi': 120,
})
```

```python
# Классический метод Рунге — Кутты 4-го порядка
# Функция работает и с одним вектором, и с массивом независимых задач
def rk4(rhs, t, z0):
    t = np.asarray(t, dtype=float)
    z0 = np.asarray(z0, dtype=float)
    z = np.zeros((len(t),) + z0.shape)
    z[0] = z0

    for j in range(len(t) - 1):
        h = t[j + 1] - t[j]
        a = rhs(t[j], z[j])
        b = rhs(t[j] + h/2, z[j] + h*a/2)
        c = rhs(t[j] + h/2, z[j] + h*b/2)
        d = rhs(t[j] + h, z[j] + h*c)

        z[j+1] = z[j] + h*(a + 2*b + 2*c + d)/6

    return z
```

```python
l = 1.0
m0 = 1.0
g = 9.81
b = k/50
F = m/10
c = b/m0
omega0 = np.sqrt(g/l)
alpha = np.linspace(0, 5, 101)

def pendulum_rhs(t, z):
    # Каждая строка z содержит [theta, omega] для своей частоты.
    return np.stack((z[:, 1],
        -c*z[:, 1] - (g/l)*np.sin(z[:, 0]) + F*np.sin(alpha*t)), axis=1)

def solve_pendulum(h):
    t = np.linspace(0, 100, round(100/h)+1)
    z0 = np.zeros((len(alpha), 2))
    z0[:, 0] = np.pi/4

    return t, rk4(pendulum_rhs, t, z0)

t, z = solve_pendulum(0.01)
t_ref, z_ref = solve_pendulum(0.005)
print(f'Частот: {len(alpha)}; b={b}; F={F}; omega0={omega0:.9f}')
```

**Вычисляем амплитуды и проверяем выход на режим**. Ниже приведены все операции: формирование матрицы четырёх гармоник, нахождение коэффициентов и восстановление полного размаха. Число обусловленности контролирует, не усиливаются ли ошибки при таком восстановлении.


```python
# Оценка амплитуды установившихся колебаний
def amplitudes(t, z, start):
    amp = np.zeros(len(alpha))
    residual = np.zeros(len(alpha))
    condition = np.zeros(len(alpha))
    harmonic_ratio = np.zeros(len(alpha))

    # Полный период для восстановления колебаний
    phase = np.linspace(0, 2 * np.pi, 4001)

    B = np.column_stack((
        np.sin(phase),
        np.cos(phase),
        np.sin(3 * phase),
        np.cos(3 * phase),
    ))

    # Участок после затухания переходного процесса
    mask = t >= start
    tt = t[mask]

    # При alpha[0] = 0 установившаяся амплитуда равна нулю
    for j in range(1, len(alpha)):
        a = alpha[j]
        theta = z[mask, j, 0]

        M = np.column_stack((
            np.sin(a * tt),
            np.cos(a * tt),
            np.sin(3 * a * tt),
            np.cos(3 * a * tt),
        ))

        # Коэффициенты первой и третьей гармоник
        q = np.linalg.lstsq(M, theta, rcond=None)[0]

        restored = B @ q

        amp[j] = (restored.max() - restored.min()) / 2
        residual[j] = np.max(np.abs(M @ q - theta))
        condition[j] = np.linalg.cond(M)

        first_harmonic = np.hypot(q[0], q[1])
        third_harmonic = np.hypot(q[2], q[3])

        harmonic_ratio[j] = third_harmonic / first_harmonic

    return amp, residual, condition, harmonic_ratio


# Амплитуды на разных временных участках и шагах интегрирования
amp, residual, condition, ratio = amplitudes(t, z, 60)
amp70, _, _, _ = amplitudes(t, z, 70)
amp_ref, _, _, _ = amplitudes(t_ref, z_ref, 60)

window_error = np.max(np.abs(amp - amp70))
step_error = np.max(np.abs(amp - amp_ref))


# Проверка повторяемости через период
period_error = np.full(len(alpha), np.nan)

for j in range(1, len(alpha)):
    period = 2 * np.pi / alpha[j]

    # Проверяем частоты, для которых период помещается в [60, 100]
    if period < 40:
        tt = np.linspace(60, 100 - period, 1501)
        theta = z[:, j, 0]

        before = np.interp(tt, t, theta)
        after = np.interp(tt + period, t, theta)

        period_error[j] = np.max(np.abs(after - before))

max_period_error = np.nanmax(period_error)


# Результаты проверки точности
print('Проверка точности')
print(
    f'Максимальная невязка гармонического описания: '
    f'{residual.max():.3e} рад'
)
print(
    f'Изменение окна [60, 100] → [70, 100]: '
    f'{window_error:.3e} рад'
)
print(
    f'Уменьшение шага 0.01 → 0.005: '
    f'{step_error:.3e} рад'
)
print(
    f'Повторяемость через период: '
    f'{max_period_error:.3e} рад'
)
print(
    f'Наибольшее число обусловленности: '
    f'{condition.max():.3f}'
)
print(
    f'Наибольшее отношение третьей гармоники к первой: '
    f'{ratio.max():.3e}'
)

assert residual.max() < 1e-4
assert window_error < 1e-4
assert step_error < 1e-6
assert max_period_error < 1e-4
assert condition.max() < 10
assert ratio.max() < 1e-3


# Амплитуда и резонансная частота линейной модели
A_linear = F / np.sqrt(
    (omega0**2 - alpha**2)**2
    + c**2 * alpha**2
)

# При alpha = 0 внешняя сила отсутствует
A_linear[0] = 0

alpha_res_linear = np.sqrt(omega0**2 - c**2 / 2)

# Максимум амплитуды на заданной сетке
jmax = np.argmax(amp)

print('\nРезультаты')
print(f'Частота максимума на сетке: {alpha[jmax]:.2f} рад/с')
print(f'Максимальная амплитуда:     {amp[jmax]:.9f} рад')
print(f'Амплитуда в градусах:       {np.degrees(amp[jmax]):.6f}°')
print(f'Линейная резонансная частота: {alpha_res_linear:.9f} рад/с')


# Сравнение нелинейной и линейной моделей
print('\nСравнение амплитуд')
print(f'{"α, рад/с":>10} {"A, рад":>14} {"A_linear, рад":>16}')

for j in [0, 1, 20, 40, 60, 61, 62, 63, 64, 80, 100]:
    print(
        f'{alpha[j]:10.2f} '
        f'{amp[j]:14.9f} '
        f'{A_linear[j]:16.9f}'
    )
```
