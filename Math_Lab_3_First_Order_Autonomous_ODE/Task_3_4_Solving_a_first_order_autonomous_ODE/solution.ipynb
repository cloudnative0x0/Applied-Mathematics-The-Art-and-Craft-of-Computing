## Решение автономного ОДУ первого порядка


## Решение
Решим уравнение
$$\frac{dx}{dt}=m\sin x+k\cos x$$
и построить $x(t)$ при нескольких начальных условиях.

**Подставим постоянные и объединяем синус с косинусом**.
$$x'=2\sin x+17\cos x.$$

Найдёс такие $R>0$ и $\varphi$, чтобы
$$2\sin x+17\cos x=R\sin(x+\varphi).$$

Используем формулу суммы комплексных числах:
$$R\sin(x+\varphi)=R\cos\varphi\sin x+R\sin\varphi\cos x.$$

Сравниваем коэффициенты:
$$R\cos\varphi=2,\qquad R\sin\varphi=17.$$

Возводим в квадрат и складываем:
$$R^2(\cos^2\varphi+\sin^2\varphi)=2^2+17^2=293,$$
$$R=\sqrt{293}\approx17.117242769.$$

Оба коэффициента положительны, поэтому $\varphi$ в первой четверти:
$$\varphi=\arctan\frac{17}{2}\approx1.453687582.$$

Получили
$$x'=R\sin(x+\varphi).$$

**Найдём стационарные решения**. До деления на правую часть проверяем, где она равна нулю:
$$\sin(x+\varphi)=0\quad\Longrightarrow\quad x+\varphi=n\pi.$$

Следовательно,
$$x(t)\equiv n\pi-\varphi,\qquad n\in\mathbb Z.$$
Это отдельные постоянные решения, которые можно потерять при разделении переменных.

**Определим устойчивость**. Пусть $f(x)=R\sin(x+\varphi)$. Тогда
$$f'(x)=R\cos(x+\varphi),\qquad f'(n\pi-\varphi)=R(-1)^n.$$

При чётном $n$ производная положительна: решения уходят от равновесия. При нечётном $n$ производная отрицательна: решения приближаются к нему.

Таким образом,
$$x_u=-\varphi+2\pi j\quad\text{— неустойчивые равновесия},$$
$$x_s=\pi-\varphi+2\pi j\quad\text{— асимптотически устойчивые равновесия}.$$
В промежутке между $x_u$ и $x_s$ правая часть положительна, а между $x_s$ и следующим $x_u$ — отрицательна. Решения с обеих сторон движутся к $x_s$.

**Решим разделением переменных**. На промежутке между соседними неустойчивыми равновесиями выбираем
$$x_s=\pi-\varphi+2\pi j,\qquad v=x-x_s,\qquad-\pi<v<\pi.$$

Тогда
$$x+\varphi=\pi+2\pi j+v,$$
$$\sin(x+\varphi)=\sin(\pi+v)=-\sin v.$$

Следовательно,
$$v'=-R\sin v.$$

При $v\ne0$ разделяем переменные:
$$\frac{dv}{\sin v}=-R\,dt.$$

Введёс интегрирование через замену $u=\tan(v/2)$:
$$\sin v=\frac{2u}{1+u^2},\qquad dv=\frac{2,du}{1+u^2}.$$

Тогда
$$\frac{dv}{\sin v}=\frac{2,du}{1+u^2}\frac{1+u^2}{2u}=\frac{du}{u},$$
$$\ln|u|=-Rt+C.$$

После возведения в экспоненту знак включаем в вещественную постоянную:
$$u=C_1e^{-Rt}.$$

Начальное условие $x(0)=x_0$ даёт
$$C_1=\tan\frac{x_0-x_s}{2}.$$

Поскольку $v\in(-\pi,\pi)$, имеем $v/2\in(-\pi/2,\pi/2)$: можно использовать главную ветвь арктангенса без скачка на $2\pi$.

**Запишем решение задачи Коши**. Для
$$-\varphi+2\pi j<x_0<2\pi-\varphi+2\pi j$$
получаем
$$x(t)=x_s+2\arctan\left[\tan\frac{x_0-x_s}{2}\,e^{-\sqrt{293}\,t}\right],\quad x_s=\pi-\varphi+2\pi j.$$
Формула также даёт постоянное решение при $x_0=x_s$. На границах промежутка стоят неустойчивые постоянные решения; их записываем отдельно.

При $t\to+\infty$ экспонента стремится к нулю, поэтому
$$\boxed{x(t)\longrightarrow x_s.}$$
Устойчивое равновесие не достигается за конечное время непостоянным решением. Правая часть гладкая, и единственность не позволяет ему пересечь равновесное решение.

**Проверяем формулу подстановкой**. Обозначим $u=C_1e^{-Rt}$. Тогда
$$x'=-\frac{2Ru}{1+u^2}.$$

С другой стороны,
$$R\sin(x+\varphi)=-R\sin(2\arctan u)=-\frac{2Ru}{1+u^2}.$$

Обе части совпадают. При $t=0$ арктангенс восстанавливает $(x_0-x_s)/2$, следовательно, начальное условие также выполнено.

**Выбираем начальные условия и время**.
Для наглядности берём один промежуток:
$$x_L=-\varphi\approx-1.453687582,$$
$$x_s=\pi-\varphi\approx1.687905071,$$
$$x_R=2\pi-\varphi\approx4.829497725.$$

Внутри промежутка выбираем пять начальных точек с долями $0.08,0.25,0.50,0.75,0.92$ от его длины. Это показывает возрастание снизу, убывание сверху и постоянное решение посередине. Отдельно отмечаем оба неустойчивых равновесия.

Характерное время малых отклонений $1/R\approx0.05842$. Отрезок $0\le t\le0.7$ содержит примерно 12 таких времён, поэтому приближение к равновесию хорошо видно. Основная сетка имеет шаг 0.001 с. Численное решение сравниваем с точной формулой.

```python
import numpy as np
import matplotlib.pyplot as plt

%matplotlib inline
%config InlineBackend.figure_format = 'retina'

plt.close('all')

# Постоянные
m = 2
k = 17

# Палитра
brown       = [0.4627, 0.3294, 0.2314]  # коричневый
light_brown = [0.7451, 0.5843, 0.4353]  # светло-коричневый
beige       = [0.8549, 0.7765, 0.6667]  # бежевый
gray        = [0.8000, 0.8000, 0.8000]  # серая сетка
background  = [1.0000, 1.0000, 1.0000]  # белый фон

# Оформление графиков
plt.rcParams.update({
    'figure.facecolor': background,
    'axes.facecolor': background,
    'savefig.facecolor': background,
    'grid.color': gray,
    'grid.alpha': 1,
    'axes.grid': True,
    'font.size': 11,
    'axes.prop_cycle': plt.cycler(
        color=[brown, light_brown, gray]
    ),
    'figure.figsize': (10, 4),
    'figure.dpi': 120,
})
```

```python
# Классический метод Рунге — Кутты 4-го порядка из семинара 14 октября.
# Функция работает и с одним вектором, и с массивом независимых задач.
def rk4(rhs, t, z0):
    t = np.asarray(t, dtype=float)
    z0 = np.asarray(z0, dtype=float)
    z = np.zeros((len(t),) + z0.shape)
    z[0] = z0
    for j in range(len(t) - 1):
        h = t[j + 1] - t[j]
        a = rhs(t[j], z[j])
        b = rhs(t[j] + h/2, z[j] + h*a/2)
        c = rhs(t[j] + h/2, z[j] + h*b/2)
        d = rhs(t[j] + h, z[j] + h*c)
        z[j+1] = z[j] + h*(a + 2*b + 2*c + d)/6
    return z
```

```python
R = np.hypot(m, k)
phi = np.arctan2(k, m)

x_left = -phi
x_stable = np.pi-phi
x_right = 2*np.pi-phi

fractions = np.array([0.08, 0.25, 0.50, 0.75, 0.92])
x0 = x_left+2*np.pi*fractions
t = np.linspace(0, 0.7, 701)

# Все пять начальных условий интегрируются независимо.
rhs = lambda t, x: m*np.sin(x)+k*np.cos(x)
x_num = rk4(rhs, t, x0)
x_exact = x_stable+2*np.arctan(
    np.tan((x0[None,:]-x_stable)/2)*np.exp(-R*t[:,None]))

error = np.max(np.abs(x_num-x_exact))

print('Параметры и контроль точности')
print('─' * 38)
print(f'R = {R:.12f},   φ = {phi:.12f}')
print(f'Неустойчивые равновесия: {x_left:.9f}; {x_right:.9f}')
print(f'Устойчивое равновесие:   {x_stable:.9f}')
print(f'Характерное время 1/R:    {1/R:.6f}')
print('Начальные условия:', np.round(x0, 6))
print(f'Максимальная ошибка RK4: {error:.3e}')

assert error < 1e-7
assert np.max(np.abs(x_exact[0]-x0)) < 1e-12

fig, axes = plt.subplots(1, 2, figsize=(13, 4.8), constrained_layout=True)
fig.suptitle('Решение и численная проверка',
             fontsize=14, fontweight='bold', color=brown)
colors = [brown, light_brown, brown, light_brown, brown]
styles = ['-', '-', ':', '--', '--']

for j in range(len(x0)):
    axes[0].plot(t, x_exact[:,j], styles[j], color=colors[j],
                 linewidth=2.0, label=f'$x(0)={x0[j]:.3f}$')
    axes[1].semilogy(t, np.maximum(np.abs(x_num[:,j]-x_exact[:,j]),1e-16),
                    styles[j], color=colors[j], label=f'x(0)={x0[j]:.3f}')

axes[0].axhline(x_left, color=gray, linestyle='--', linewidth=1.2)
axes[0].axhline(x_right, color=gray, linestyle='--', linewidth=1.2)
axes[0].axhline(x_stable, color=beige, linestyle=':', linewidth=2.0,
                    label='устойчивое равновесие')
axes[0].annotate('$x_s$', xy=(0.67, x_stable), xytext=(0.62, x_stable + 0.38),
                 color=brown, arrowprops={'arrowstyle': '-', 'color': brown})
axes[0].set(title='Точные решения $x(t)$', xlabel='$t$', ylabel='$x(t)$',
            xlim=(t[0], t[-1]))
axes[0].legend(fontsize=8, ncol=2, frameon=False, loc='lower right')
axes[1].set(title='Ошибка Рунге относительно точной формулы',
            xlabel='$t$', ylabel='Абсолютная ошибка', xlim=(t[0], t[-1]))

axes[1].legend(fontsize=8, frameon=False, loc='upper right')

for ax in axes:
    ax.spines[['top', 'right']].set_visible(False)
    ax.grid(True, linewidth=0.7)

plt.show()
```

**Комментарий к графикам**. Левый график. Построены решения из интервала между двумя неустойчивыми равновесиями $x_L=-1.453687582$ и $x_R=4.829497725$. Устойчивое равновесие в этом интервале равно $x_s=1.687905071$ и отмечено точечной горизонтальной линией.

Для начальных условий $x(0)=-0.951033$ и $x(0)=0.117109$, расположенных ниже $x_s$, правая часть уравнения положительна, поэтому решения монотонно возрастают к $x_s$. Для $x(0)=3.258701$ и $x(0)=4.326843$, расположенных выше $x_s$, правая часть отрицательна, поэтому решения монотонно убывают к $x_s$. При $x(0)=x_s$ получается постоянное решение $x(t)\equiv x_s$.

Пунктирные горизонтальные линии $x_L$ и $x_R$ — неустойчивые постоянные решения. Непостоянная траектория не пересекает их и не достигает $x_s$ за конечное время; она только приближается к $x_s$ при $t\to+\infty$.

Правый график. Показана абсолютная разность между значениями, полученными методом RK4, и точной формулой. На всей сетке максимальная ошибка равна $5.67\cdot10^{-10}$, поэтому численный расчёт согласуется с аналитическим решением.

Пусть $\varphi=\arctan(17/2)$ и $R=\sqrt{293}$. Если начальное значение не совпадает с неустойчивым равновесием, выбираем $j=\lfloor(x_0+\varphi)/(2\pi)\rfloor$ и $x_s=\pi-\varphi+2\pi j$. Тогда
\(x(t)=x_s+2\arctan\!\left(\tan\frac{x_0-x_s}{2}\,e^{-\sqrt{293}t}\right),\)
причём $x(t)\to x_s$ при $t\to+\infty$. Кроме того, $x(t)\equiv n\pi-\varphi$, $n\in\mathbb Z$, — все постоянные решения; среди них решения с нечётным $n$ устойчивы, с чётным $n$ неустойчивы.
