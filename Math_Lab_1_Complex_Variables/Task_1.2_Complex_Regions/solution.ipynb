### Решение
Комплексным числом назовем выражение вида $z=a+bi$, где $a$ – вещественная часть, $b$ – мнимая часть. На комплексной плоскости числу $z=a+bi$ соответствует точка с координатами $(a,b)$.


#### Первая область

Первое область задачи имеет вид $|z+mi|>k$. Подставим $z=a+bi$ в выражение под модулем, имеем: $z+mi=a+bi+mi$. Вынесем общий множитель за скобки:
$$
z+mi=a+i(b+m).
$$

Следующим шагом применим формулу модуля комплексного числа.
$$
|z|=\sqrt{a^2+b^2}.
$$

В полученном числе вещественная часть равна $a$, а мнимая часть равна $b+m$. Поэтому
$$
|z+mi|=\sqrt{a^2+(b+m)^2}.
$$

Тогда первое условие принимает вид:
$$
\sqrt{a^2+(b+m)^2}>k.
$$

Так как обе части положительные, возьмем их в квадрат, чтобы избавиться от корня в левой части:

$$
a^2+(b+m)^2>k^2.
$$


Далее нам нужно понять, какую область задаёт неравенство $|z+mi|>k$. Известно, что $|z-w|$ – расстояние между переменной точкой $z$ и фиксированной точкой $w$. Если найти такое $w$, что $z+mi=z-w$, то можно узнать: центр окружности, ее радиус и какая сторона этой окружности удовлетворяет условию первой области. Цель: определить положение области на комплексной плоскости.

Найдем $w$, которое будет удовлетворять равенство $z+mi=z-w$.
Первым действием вычтем $z$ из обеих частей, имеем: $mi=-w$, умножим обе части равенства на $-1$, получим: $w=-mi$. Таким образом мы нашли искомое число. Подставим это выражение в исходное, получим:
$$
|z+mi|=|z-(-mi)|
$$
Таким образом мы выяснили, что расстояние измеряется от точки $z$ до точки $-mi$.

Рассмотрим найденное комплексное число $-mi$. Можно заметить, что коэффициент данного числа является мнимой частью, вещественная часть отсутствует, следовательно можно сделать сквозное заключение, что числу при $w=-mi$ на комплексной плоскости соответствует точка с координатами $(0,-m)$.

Следующем действием выразим это расстояние через вещественную часть $a$ и мнимую часть $b$ переменной точки $z$.
Расстояние между комплексными числами задаётся формулой:
$$
|z-w|=\sqrt{(a-c)^2+(b-d)^2}.
$$

Рассматриваемые точки в формуле расстояния у нас следующие:
$$
a=a,\qquad b=b, \qquad c=0,\qquad d=-m.
$$

Подставим точки в формулу расстояния:
$$
|z-(-mi)|=\sqrt{(a-0)^2+(b-(-m))^2}.
$$

Оба выражения под корнем показывают разности вещественной и мнимой частей: $a-0=a$ и $b-(-m)=b+m$. Имеем:
$$
|z-(-mi)|=\sqrt{a^2+(b+m)^2}.
$$

Ранее было установлено, что $|z+mi|=|z-(-mi)|$, подставим полученное выражение из формулы расстояния в данное равенство:
$$
|z+mi|=\sqrt{a^2+(b+m)^2}.
$$

То есть мы заменили модуль комплексного числа выражением, содержащим только вещественные числа $a$, $b$ и $m$, чтобы в дальнейшем вычислять их в программе.

Вернемся к исходному условию первой области $|z+mi|>k$, вместо $|z+mi|$ подставим найденное выражение:
$$
\sqrt{a^2+(b+m)^2}>k.
$$

Обе части положительные, имеем право избавиться от корня в левой части:
$$
a^2+(b+m)^2>k^2.
$$

В итоге мы получили условие первой области через вещественные координаты точки $z$.

Далее определим границу области. Граница отделяет точки, удовлетворяющие неравенству, от точек, которые ему не удовлетворяют.
На границе расстояние от точки $z$ до точки $(0,-m)$ должно быть в точности равно $k$. Поэтому приравняем обе части, имеем:
$$
a^2+(b+m)^2=k^2.
$$

Это равенство означает, что расстояние от точки $(a,b)$ до точки $(0,-m)$ равно $k$.

Исходя из лекции о комплексных числах, можем сделать сквозное заключение о том, что при $|z|>1$ расстояние до начала координат больше радиуса единичной окружности, а соответствующие точки расположены снаружи окружности.
