## Пересечение областей на комплексной плоскости

**Задача 1.2.** Изобразить с помощью компьютера пересечение областей $|z+m\cdot i|>k$ и $|z-k\cdot i|<m.$ Рисунок должна сопровождать легенда (какая область изображена). Прямоугольник, на котором строится область, нужно выбрать самостоятельно. Нужно объяснить этот свой выбор. Программный код нужно приложить.

Муравьёв Герман

```python
import numpy as np
import matplotlib.pyplot as plt

from matplotlib.lines import Line2D
from matplotlib.patches import Patch

%config InlineBackend.figure_format = 'retina'

# Постоянные
m = 2
k = 17

# Палитра
dark       = [0.0039, 0.0275, 0.2118] # граница первой области, ось и отмеченные точки
navy       = [0.0510, 0.1098, 0.2588] # граница второй области, сетка и рамка легенды
blue       = [0.1333, 0.2235, 0.4353] # пересечение двух областей
background = [1.0000, 1.0000, 1.0000]
```

### Решение
Комплексным числом назовем выражение вида $z=a+bi$, где $a$ – вещественная часть, $b$ – мнимая часть. На комплексной плоскости числу $z=a+bi$ соответствует точка с координатами $(a,b)$.


#### Первая область

Первое область задачи имеет вид $|z+mi|>k$. Подставим $z=a+bi$ в выражение под модулем, имеем: $z+mi=a+bi+mi$. Вынесем общий множитель за скобки:
$$
z+mi=a+i(b+m).
$$

Следующим шагом применим формулу модуля комплексного числа.
$$
|z|=\sqrt{a^2+b^2}.
$$

В полученном числе вещественная часть равна $a$, а мнимая часть равна $b+m$. Поэтому
$$
|z+mi|=\sqrt{a^2+(b+m)^2}.
$$

Тогда первое условие принимает вид:
$$
\sqrt{a^2+(b+m)^2}>k.
$$

Так как обе части положительные, возьмем их в квадрат, чтобы избавиться от корня в левой части:

$$
a^2+(b+m)^2>k^2.
$$


Далее нам нужно понять, какую область задаёт неравенство $|z+mi|>k$. Известно, что $|z-w|$ – расстояние между переменной точкой $z$ и фиксированной точкой $w$. Если найти такое $w$, что $z+mi=z-w$, то можно узнать: центр окружности, ее радиус и какая сторона этой окружности удовлетворяет условию первой области. Цель: определить положение области на комплексной плоскости.

Найдем $w$, которое будет удовлетворять равенство $z+mi=z-w$.
Первым действием вычтем $z$ из обеих частей, имеем: $mi=-w$, умножим обе части равенства на $-1$, получим: $w=-mi$. Таким образом мы нашли искомое число. Подставим это выражение в исходное, получим:
$$
|z+mi|=|z-(-mi)|
$$
Таким образом мы выяснили, что расстояние измеряется от точки $z$ до точки $-mi$.

Рассмотрим найденное комплексное число $-mi$. Можно заметить, что коэффициент данного числа является мнимой частью, вещественная часть отсутствует, следовательно можно сделать сквозное заключение, что числу при $w=-mi$ на комплексной плоскости соответствует точка с координатами $(0,-m)$.

Следующем действием выразим это расстояние через вещественную часть $a$ и мнимую часть $b$ переменной точки $z$.
Расстояние между комплексными числами задаётся формулой:
$$
|z-w|=\sqrt{(a-c)^2+(b-d)^2}.
$$

Рассматриваемые точки в формуле расстояния у нас следующие:
$$
a=a,\qquad b=b, \qquad c=0,\qquad d=-m.
$$

Подставим точки в формулу расстояния:
$$
|z-(-mi)|=\sqrt{(a-0)^2+(b-(-m))^2}.
$$

Оба выражения под корнем показывают разности вещественной и мнимой частей: $a-0=a$ и $b-(-m)=b+m$. Имеем:
$$
|z-(-mi)|=\sqrt{a^2+(b+m)^2}.
$$

Ранее было установлено, что $|z+mi|=|z-(-mi)|$, подставим полученное выражение из формулы расстояния в данное равенство:
$$
|z+mi|=\sqrt{a^2+(b+m)^2}.
$$

То есть мы заменили модуль комплексного числа выражением, содержащим только вещественные числа $a$, $b$ и $m$, чтобы в дальнейшем вычислять их в программе.

Вернемся к исходному условию первой области $|z+mi|>k$, вместо $|z+mi|$ подставим найденное выражение:
$$
\sqrt{a^2+(b+m)^2}>k.
$$

Обе части положительные, имеем право избавиться от корня в левой части:
$$
a^2+(b+m)^2>k^2.
$$

В итоге мы получили условие первой области через вещественные координаты точки $z$.

Далее определим границу области. Граница отделяет точки, удовлетворяющие неравенству, от точек, которые ему не удовлетворяют.
На границе расстояние от точки $z$ до точки $(0,-m)$ должно быть в точности равно $k$. Поэтому приравняем обе части, имеем:
$$
a^2+(b+m)^2=k^2.
$$

Это равенство означает, что расстояние от точки $(a,b)$ до точки $(0,-m)$ равно $k$.

Исходя из лекции о комплексных числах, можем сделать сквозное заключение о том, что при $|z|>1$ расстояние до начала координат больше радиуса единичной окружности, а соответствующие точки расположены снаружи окружности.

#### Вторая область

Вторая область задачи имеет вид $|z-ki|<m$. Подставим $z=a+bi$ в выражение под модулем, имеем: $z-ki=a+bi-ki$. Вынесем общий множитель за скобки:
$$
z-ki=a+i(b-k).
$$

Следующим шагом применим формулу модуля комплексного числа.
$$
|z|=\sqrt{a^2+b^2}.
$$

В полученном числе вещественная часть равна $a$, а мнимая часть равна $b-k$. Поэтому
$$
|z-ki|=\sqrt{a^2+(b-k)^2}.
$$

Тогда второе условие принимает вид:
$$
\sqrt{a^2+(b-k)^2}>k.
$$

Так как левая часть неотрицательна, а $m>0$, возьмем обе части в квадрат, чтобы избавиться от корня в левой части:
$$
a^2+(b-k)^2>k^2.
$$



Далее нам нужно понять, какую область задаёт неравенство $|z-ki|<m$. Известно, что $|z-w|$ – расстояние между переменной точкой $z$ и фиксированной точкой $w$. Если найти такое $w$, что $z-ki=z-w$, то можно узнать: центр окружности, ее радиус и какая сторона этой окружности удовлетворяет условию второй области. Цель: определить положение области на комплексной плоскости.

Найдем $w$, которое будет удовлетворять равенство $z-ki=z-w$.
Первым действием вычтем $z$ из обеих частей, имеем: $-ki=-w$, умножим обе части равенства на $-1$, получим: $w=ki$. Таким образом мы нашли искомое число. Подставим это выражение в исходное, получим:
$$
|z-ki|=|z-ki|
$$
Таким образом мы выяснили, что расстояние измеряется от точки $z$ до точки $ki$.

Рассмотрим найденное комплексное число $ki$. Можно заметить, что коэффициент данного числа является мнимой частью, вещественная часть отсутствует, следовательно можно сделать сквозное заключение, что числу при $w=ki$ на комплексной плоскости соответствует точка с координатами $(0, k)$.

Следующем действием выразим это расстояние через вещественную часть $a$ и мнимую часть $b$ переменной точки $z$.
Расстояние между комплексными числами задаётся формулой:
$$
|z-w|=\sqrt{(a-c)^2+(b-d)^2}.
$$

Рассматриваемые точки в формуле расстояния у нас следующие:
$$
a=a,\qquad b=b,\qquad c=0,\qquad d=k.
$$

Подставим точки в формулу расстояния:
$$
|z-ki|=\sqrt{(a-0)^2+(b-k)^2}.
$$

Оба выражения под корнем показывают разности вещественной и мнимой частей: $a-0=a$ и $b-k=b-k$$. Имеем:
$$
|z-ki|=\sqrt{a^2+(b-k)^2}.
$$

Ранее было установлено, что $|z-ki|=|z-ki|$, подставим полученное выражение из формулы расстояния в данное равенство:
$$
|z-ki|=\sqrt{a^2+(b-k)^2}.
$$

То есть мы заменили модуль комплексного числа выражением, содержащим только вещественные числа $a$, $b$ и $m$, чтобы в дальнейшем вычислять их в программе.

Вернемся к исходному условию первой области $|z-ki|<m$, вместо $|z-ki|$ подставим найденное выражение:
$$
\sqrt{a^2+(b-k)^2}<m.
$$

Обе части положительные, имеем право избавиться от корня в левой части:
$$
a^2+(b-k)^2<m^2.
$$

В итоге мы получили условие первой области через вещественные координаты точки $z$.

Далее определим границу области. Граница отделяет точки, удовлетворяющие неравенству, от точек, которые ему не удовлетворяют.
На границе расстояние от точки $z$ до точки $(0,k)$ должно быть в точности равно $m$. Поэтому приравняем обе части, имеем:
$$
a^2+(b-k)^2=m^2.
$$

Это равенство означает, что расстояние от точки $(a,b)$ до точки $(0,k)$ равно $m$.
Исходя из лекции о комплексных числах, можем сделать сквозное заключение о том, что при $|z|<1$ расстояние до начала координат меньше радиуса единичной окружности, а соответствующие точки расположены внутри окружности.

#### Подставляем значения в обе области

Для первой области $|z+2i|>17$ получаем:
$$
a^2+(b+2)^2>17^2=289.
$$

Это точки, расположенные снаружи окружности радиуса $17$ с центром $(0,-2)$.

Для второй области $|z-17i|<2$ получаем:
$$
a^2+(b-17)^2<2^2=4.
$$

Это точки, расположенные внутри окружности радиуса $2$ с центром $(0,17)$. Граница окружности не входит в область, поскольку неравенство строгое.

##### Находим пересечение областей

Чтобы найти общую точку границ, нужно решить систему уравнений двух окружностей, имеем:
$$
\begin{cases}
a^2+(b+2)^2=289,\\
a^2+(b-17)^2=4.
\end{cases}
$$

Вычтем второе уравнение из первого, имеем:
$$
(b+2)^2-(b-17)^2=285.
$$

Раскроем скобки, применив формулу сокращенного умножения:
$$
(b+2)^2=b^2+4b+4,
(b-17)^2=b^2-34b+289.
$$

Имеем:
$$
b^2+4b+4-\left(b^2-34b+289\right)=285.
$$

Раскроем скобки и приведем подобные:
$$
b^2+4b+4-b^2+34b-289=285,
$$

имеем:
$$
38b=570, \quad b=15
$$

Найдем $a$, подставив уже известную переменную b в любую часть системы из двух уравнений, выберем вторую: $a^2+(b-17)^2=4.$. Имеем:
$$
a^2+(15-17)^2=4. \quad a=0.
$$

Получаем: $(a,b)=(0,15)$.

```python
# Прямоугольник построения
a_min, a_max = -4, 4
b_min, b_max = 13, 21

# Сетка на комплексной плоскости
a_values = np.linspace(a_min, a_max, 1200)
b_values = np.linspace(b_min, b_max, 1200)
A, B = np.meshgrid(a_values, b_values)

# Расстояние от точки (a, b) до точки (0, -m)
distance_first = np.sqrt(A**2 + (B + m)**2)
# Расстояние от точки (a, b) до точки (0, k)
distance_second = np.sqrt(A**2 + (B - k)**2)

# Первая область: |z + mi| > k
area_first = distance_first > k
# Вторая область: |z - ki| < m
area_second = distance_second < m
# Пересечение двух областей
area_intersection = area_first & area_second
```

```python
# Проверяем на сетке полученный аналитический результат
assert np.array_equal(area_intersection, area_second)
print("Пересечение совпадает с кругом |z - 17i| < 2.")
```

##### Находим общую точку границ
Координаты общей точки границ:
$$
(a,b)=(0,15).
$$

Найдем соответствующее комплексное число при $z=a+bi$, имеем:
$$
z=0+15i. \quad z=15i.
$$

Введем проверку на границы:
$$
|15i+2i|=|17i|=17, \quad |15i-17i|=|-2i|=2.
$$

Точка $15i$ принадлежит обеим границам. Но она не входит ни в одну из двух условных областей, потому что исходные неравенства строгие:
$$
|z+2i|>17,
\qquad
|z-17i|<2.
$$

##### Выбираем прямоугольник построения
Вторая область является кругом радиуса $2$ с центром в точке $(0,17)$. Следовательно, она расположена в пределах
$$
-2<a<2,\qquad 15<b<19.
$$
Чтобы область не соприкасалась с краями рисунка и вокруг неё оставалось свободное пространство, выберем прямоугольник:
$$
-4\leq a\leq4,\qquad 13\leq b\leq21.
$$

В выбранный прямоугольник полностью входит вторая область. Кроме того, в нём видны границы обеих областей и их общая точка $(a,b)=(0,15).$

```python
figure, axes = plt.subplots(
    figsize=(10.5, 7.0),
    facecolor=background
)

axes.set_facecolor(background)

# Закрашиваем пересечение двух областей
axes.contourf(
    A,
    B,
    area_intersection.astype(int),
    levels=[0.5, 1.5],
    colors=[blue],
    alpha=0.88,
)

# Граница первой области:
# sqrt(a^2 + (b + m)^2) = k
axes.contour(
    A,
    B,
    distance_first,
    levels=[k],
    colors=[dark],
    linestyles=["--"],
    linewidths=[2.0],
)

# Граница второй области:
# sqrt(a^2 + (b - k)^2) = m
axes.contour(
    A,
    B,
    distance_second,
    levels=[m],
    colors=[navy],
    linestyles=["--"],
    linewidths=[2.0],
)

# Центр второй окружности (0, k)
axes.scatter(
    0.0,
    k,
    color=dark,
    marker="x",
    s=70,
    linewidth=2.0,
    zorder=5,
)

# Общая точка границ (0, k - m)
axes.scatter(
    0.0,
    k - m,
    facecolor=background,
    edgecolor=dark,
    s=75,
    linewidth=2.0,
    zorder=6,
)

# Элементы легенды
legend_handles = [
    Patch(
        facecolor=blue,
        edgecolor=blue,
        alpha=0.88,
        label=r"Пересечение: $|z+2i|>17$ и $|z-17i|<2$",
    ),
    Line2D(
        [0],
        [0],
        color=dark,
        linestyle="--",
        linewidth=2.0,
        label=r"Граница $|z+2i|=17$ (не входит)",
    ),
    Line2D(
        [0],
        [0],
        color=navy,
        linestyle="--",
        linewidth=2.0,
        label=r"Граница $|z-17i|=2$ (не входит)",
    ),
    Line2D(
        [0],
        [0],
        color=dark,
        marker="x",
        linestyle="None",
        markersize=8,
        markeredgewidth=2.0,
        label=r"Центр второй окружности $(0,17)$",
    ),
    Line2D(
        [0],
        [0],
        color=dark,
        marker="o",
        markerfacecolor=background,
        linestyle="None",
        markersize=8,
        markeredgewidth=2.0,
        label=r"Общая точка границ $(0,15)$ (не входит)",
    ),
]

# Мнимая ось
axes.axvline(
    0.0,
    color=dark,
    linewidth=0.8
)

# Границы выбранного прямоугольника
axes.set_xlim(a_min, a_max)
axes.set_ylim(b_min, b_max)

# Одинаковый масштаб по обеим осям
axes.set_aspect("equal", adjustable="box")

# Подписи осей и заголовок
axes.set_xlabel(r"$a=\operatorname{Re}z$")
axes.set_ylabel(r"$b=\operatorname{Im}z$")
axes.set_title("Пересечение областей на комплексной плоскости")

# Координатная сетка
axes.grid(
    color=navy,
    alpha=0.18
)

# Легенда
axes.legend(
    handles=legend_handles,
    loc="upper center",
    bbox_to_anchor=(0.5, -0.13),
    frameon=True,
    facecolor=background,
    edgecolor=navy,
    fontsize=9,
)

# Оставляем место под легенду
figure.subplots_adjust(bottom=0.30)

plt.show()
```

##### Вывод

Искомая область – круг радиуса $2$ с центром в точке $17i$, то есть в точке $(0, 17)$ комплексной плоскости. Его граница не входит в область. Первая и вторая границы имеют общую точку $15i$, но эта точка также не входит в пересечение.

На рисунке искомое пересечение закрашено синим цветом. Пунктиром показаны границы $|z+2i|=17$ и $|z-17i|=2$.
