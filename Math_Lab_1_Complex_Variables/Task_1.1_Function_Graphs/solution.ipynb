Для перехода к построению графиков, сначала вспомним, что мнимая единица `i` определяется равенством: $i^2=-1.$

Комплексным числом называется число вида: $z=a+bi,$ где `a` и `b` — вещественные числа.

При этом, число `a` называется вещественной частью `z`, откуда: $\operatorname{Re}z=a.$

Число `b` называется мнимой частью `z`, откуда: $\operatorname{Im}z=b.$

---
#### Комплекснозначная функция
Комплекснозначная функция вещественной переменной сопоставляет каждому $x\in\mathbb R$ комплексное число:

$$
x\longmapsto f(x)\in\mathbb C.
$$

Её можно записать в виде

$$
f(x)=u(x)+iv(x),
$$

где $u(x)$ и $v(x)$ — вещественные функции: $u(x)=\operatorname{Re}f(x), v(x)=\operatorname{Im}f(x)$ => $f(x)=\operatorname{Re}f(x)+i\operatorname{Im}f(x).$

Одним обычным графиком $y=f(x)$ комплекснозначную функцию представить нельзя, поскольку каждому $x$ соответствуют две вещественные координаты: $u(x)=\operatorname{Re}f(x), v(x)=\operatorname{Im}f(x).$

Поэтому для каждой комплекснозначной функции отдельно строятся графики её вещественной и мнимой частей. В данной задаче необходимо получить четыре вещественные функции: $\operatorname{Re}f(x), \operatorname{Im}f(x), \operatorname{Re}g(x), \operatorname{Im}g(x).$

---
#### Нахождение вещественной и мнимой частей

Представить каждую комплекснозначную функцию в виде
$$
u(x)+iv(x),
$$

поскольку тогда выражение без $i$ будет вещественной частью, а коэффициент при $i$ — мнимой частью.

---
#### Комплексная экспонента
В показателе экспоненты содержится как вещественная, так и мнимая часть:

$$
(m+ki)x=mx+ikx.
$$

Используя свойство экспоненты $e^{a+b}=e^ae^b$, разделим их:

$$
e^{(m+ki)x}=e^{mx}e^{ikx}.
$$

Множитель $e^{mx}$ вещественный, значит проблема сводится к преобразованию экспоненты с мнимым показателем $e^{ikx}$.

Исходя из текущего смысла, применим формулу Эйлера (формула Эйлера не только позволяет выделить вещественную и мнимую части, но и показывает общее устройство комплексной экспоненты: её модуль изменяется как $e^{mx}$, а угол – как $kx$: $\left|e^{(m+ki)x}\right|=e^{mx}$, поскольку $\cos^2(kx)+\sin^2(kx)=1$), так как её правая часть уже имеет стандартный вид комплексного числа $a+ib$:

$$
e^{i\varphi}=\cos\varphi+i\sin\varphi.
$$

Из формулы Эйлера следуют равенства

$$
\operatorname{Re}\left(e^{i\varphi}\right)=\cos\varphi,
\qquad
\operatorname{Im}\left(e^{i\varphi}\right)=\sin\varphi.
$$

Положим $\varphi=kx$. Тогда

$$
e^{ikx}=\cos(kx)+i\sin(kx).
$$

Подставляем это выражение в исходную функцию:

$$
e^{(m+ki)x}
=
e^{mx}e^{ikx}
=
e^{mx}\bigl(\cos(kx)+i\sin(kx)\bigr).
$$

Раскрываем скобки:

$$
e^{(m+ki)x}
=
e^{mx}\cos(kx)
+
i e^{mx}\sin(kx).
$$

Теперь выражение имеет необходимый вид $u(x)+iv(x)$. Следовательно,

$$
\operatorname{Re}f(x)=e^{mx}\cos(kx),
$$

$$
\operatorname{Im}f(x)=e^{mx}\sin(kx).
$$
