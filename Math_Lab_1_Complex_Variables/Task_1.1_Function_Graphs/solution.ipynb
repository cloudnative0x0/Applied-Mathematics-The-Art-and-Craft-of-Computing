# Графики комплекснозначных функций

**Задача 1.1.** Построить графики функций $\operatorname{Re}\{\exp[(m+k\cdot i)\cdot x]\}$, $\operatorname{Im}\{\exp[(m+k\cdot i)\cdot x]\}$, где $m$ — номер группы (1 или 2), $k$ — Ваш номер в списке группы. Нужно построить графики на компьютере, сопроводив комментариями. Кривые построить разными цветами. Рисунок должна сопровождать легенда (какая кривая изображена). Отрезок, на котором строится график, нужно выбрать самостоятельно так, чтобы из графика можно было понять, как устроена функция в целом. Нужно объяснить свой выбор. Программный код нужно приложить. То же задание для функций $\operatorname{Re}\{\cos[(m+k\cdot i)\cdot x]\}$, $\operatorname{Im}\{\cos[(m+k\cdot i)\cdot x]\}$.

Муравьёв Герман

```python
import numpy as np
import matplotlib.pyplot as plt

%config InlineBackend.figure_format = 'retina'

m = 2
k = 17

# Цветовая палитра графиков
maroon     = [0.5020, 0, 0.1255]      # #800020 — вещественная часть
red        = [0.8314, 0.3137, 0.3765] # #D45060 — мнимая часть
gray       = [0.40, 0.40, 0.40]       # огибающие
background = [1.00, 1.00, 1.00]       # чистый белый фон

# Подписи легенды для комплексной экспоненты
label_exp_re = r"Вещественная часть: $\operatorname{Re}f(x)=e^{2x}\cos(17x)$"
label_exp_im = r"Мнимая часть: $\operatorname{Im}f(x)=e^{2x}\sin(17x)$"
label_exp_envelope = r"Огибающие: $y=\pm e^{2x}$"

# Подписи легенды для комплексного косинуса
label_cos_re = (
    r"Вещественная часть: "
    r"$\operatorname{Re}g(x)=\cos(2x)\cosh(17x)$"
)
label_cos_im = (
    r"Мнимая часть: "
    r"$\operatorname{Im}g(x)=-\sin(2x)\sinh(17x)$"
)
```

Цель: исследовать две комплекснозначные функции вещественной переменной $x$ при $m=2$ и $k=17$:

$$
f(x)=\exp((m+ki)x),
\qquad
g(x)=\cos((m+ki)x).
$$

---
#### Комплекснозначные функции

Для построения графиков представим каждую комплекснозначную функцию в виде $h(x)=u(x)+iv(x),$ где

$$
u(x)=\operatorname{Re}h(x),
\qquad
v(x)=\operatorname{Im}h(x).
$$

Таким образом, необходимо найти четыре вещественные функции:

$$
\operatorname{Re}f(x),\quad
\operatorname{Im}f(x),\quad
\operatorname{Re}g(x),\quad
\operatorname{Im}g(x).
$$

---
#### Решение

##### Комплексная экспонента

Рассмотрим функцию $f(x)=e^{(m+ki)x}.$ Разделим вещественную и мнимую части показателя экспоненты:

$$
(m+ki)x=mx+ikx.
$$

Используя свойство $e^{a+b}=e^ae^b$, получаем $e^{(m+ki)x}=e^{mx}e^{ikx}.$

Из лекции «Комплексные числа» известна формула Эйлера: $e^{i\varphi}=\cos\varphi+i\sin\varphi.$

Применяя её к выражению $e^{ikx}$, получаем

$$
e^{ikx}=\cos(kx)+i\sin(kx).
$$

Следовательно,

$$
e^{(m+ki)x}
=
e^{mx}\cos(kx)+i e^{mx}\sin(kx).
$$

Выражение приведено к виду $u(x)+iv(x)$, поэтому

$$
\operatorname{Re}f(x)=e^{mx}\cos(kx),
\qquad
\operatorname{Im}f(x)=e^{mx}\sin(kx).
$$

##### Комплексный косинус

Рассмотрим функцию $g(x)=\cos((m+ki)x).$

Из формулы Эйлера, приведённой в лекции «Комплексные числа», следует представление косинуса через экспоненты. Для комплексного аргумента используем формулу

$$
\cos z=\frac{e^{iz}+e^{-iz}}{2}.
$$

Обозначим аргумент косинуса через $z=(m+ki)x.$

Тогда, учитывая $i^2=-1$, получаем

$$
iz=-kx+imx,
\qquad
-iz=kx-imx.
$$

По формуле Эйлера

$$
e^{iz}=e^{-kx}\bigl(\cos(mx)+i\sin(mx)\bigr),
$$

$$
e^{-iz}=e^{kx}\bigl(\cos(mx)-i\sin(mx)\bigr).
$$

Подставляя эти выражения в формулу косинуса и группируя вещественную и мнимую части, получаем

$$
\cos((m+ki)x)
=
\cos(mx)\frac{e^{kx}+e^{-kx}}{2}
-
i\sin(mx)\frac{e^{kx}-e^{-kx}}{2}.
$$

Введём обозначения гиперболических функций:

$$
\cosh t=\frac{e^t+e^{-t}}{2},
\qquad
\sinh t=\frac{e^t-e^{-t}}{2}.
$$

При $t=kx$ имеем

$$
\frac{e^{kx}+e^{-kx}}{2}=\cosh(kx),
\qquad
\frac{e^{kx}-e^{-kx}}{2}=\sinh(kx).
$$

Следовательно,

$$
\cos((m+ki)x)
=
\cos(mx)\cosh(kx)-i\sin(mx)\sinh(kx).
$$

Поэтому

$$
\operatorname{Re}g(x)=\cos(mx)\cosh(kx),
\qquad
\operatorname{Im}g(x)=-\sin(mx)\sinh(kx).
$$

#### Подстановка значений параметров

Подставим $m=2$ и $k=17$ в полученные формулы.

Для комплексной экспоненты:

$$
\operatorname{Re}f(x)=e^{2x}\cos(17x),
\qquad
\operatorname{Im}f(x)=e^{2x}\sin(17x).
$$

Для комплексного косинуса:

$$
\operatorname{Re}g(x)=\cos(2x)\cosh(17x),
\qquad
\operatorname{Im}g(x)=-\sin(2x)\sinh(17x).
$$

```python
def f_re(x):
    return np.exp(m * x) * np.cos(k * x)

def f_im(x):
    return np.exp(m * x) * np.sin(k * x)

def g_re(x):
    return np.cos(m * x) * np.cosh(k * x)

def g_im(x):
    return -np.sin(m * x) * np.sinh(k * x)
```

#### Проверка полученных формул

Сравним аналитические формулы с непосредственным вычислением комплексных функций в нескольких точках.

```python
def f_complex(x):
    return np.exp((m + 1j * k) * x)

def g_complex(x):
    return np.cos((m + 1j * k) * x)

x_test = np.linspace(-1.0, 1.0, 1000)

assert np.allclose(f_re(x_test), np.real(f_complex(x_test)))
assert np.allclose(f_im(x_test), np.imag(f_complex(x_test)))
assert np.allclose(g_re(x_test), np.real(g_complex(x_test)))
assert np.allclose(g_im(x_test), np.imag(g_complex(x_test)))

print("Успех.")
```

#### Выбор отрезка для комплексной экспоненты

Представим функцию в виде

$$
f(x)=e^{(2+17i)x}=e^{2x}e^{17ix}.
$$

Из лекции «Комплексные числа» известно, что в показательной форме модуль определяет расстояние до начала координат, а аргумент – угол поворота. Множитель $e^{17ix}$ имеет модуль $1$ и задаёт поворот, а множитель $e^{2x}$ изменяет модуль функции. Поэтому

$$
|f(x)|=e^{2x},
\qquad
\Delta\arg f=17\Delta x.
$$

Один полный поворот соответствует изменению аргумента на $2\pi$. Поэтому период колебательных множителей определяется равенством

$$
17T=2\pi,
$$

откуда

$$
T=\frac{2\pi}{17}\approx0.370.
$$

Поскольку вещественная и мнимая части являются координатами комплексного числа с модулем $e^{2x}$,

$$
\left|\operatorname{Re}f(x)\right|\leq e^{2x},
\qquad
\left|\operatorname{Im}f(x)\right|\leq e^{2x}.
$$

Следовательно, графики находятся между огибающими $y=-e^{2x}$ и $y=e^{2x}$.

Выберем отрезок $[-1,1]$. На нём изменение аргумента равно

$$
17\bigl(1-(-1)\bigr)=34,
$$

поэтому число полных поворотов составляет

$$
\frac{34}{2\pi}=\frac{17}{\pi}\approx5.41.
$$

Одновременно модуль изменяется от $e^{-2}\approx0.135$ до $e^2\approx7.389$. Таким образом, на отрезке $[-1,1]$ видно более пяти колебаний и изменение их размаха. Больший отрезок брать нецелесообразно: из-за быстрого роста экспоненты часть графика с малыми значениями станет плохо различимой.

```python
x_exp = np.linspace(-1.0, 1.0, 4000)

y_exp_re = f_re(x_exp)
y_exp_im = f_im(x_exp)
envelope = np.exp(m * x_exp)

figure = plt.figure(facecolor=background)
figure.set_size_inches(12.0, 6.0)
axes = figure.gca()
axes.set_facecolor(background)

axes.plot(
    x_exp,
    y_exp_re,
    color=maroon,
    linewidth=1.5,
    label=label_exp_re,
)
axes.plot(
    x_exp,
    y_exp_im,
    color=red,
    linewidth=1.5,
    label=label_exp_im,
)
axes.plot(
    x_exp,
    envelope,
    color=gray,
    linestyle="--",
    linewidth=1.0,
    label=label_exp_envelope,
)
axes.plot(
    x_exp,
    -envelope,
    color=gray,
    linestyle="--",
    linewidth=1.0,
)

axes.axhline(0.0, color="black", linewidth=0.8)
axes.axvline(0.0, color="black", linewidth=0.8)
axes.set_xlabel("$x$")
axes.set_ylabel("$y$")
axes.set_title("Вещественная и мнимая части комплексной экспоненты")
axes.legend()
axes.grid(alpha=0.3)
axes.set_xlim(-1.0, 1.0)

plt.show()
```

##### Комментарий к графику комплексной экспоненты

Исходя из получившегося графика, можно заметить, что обе функции совершают колебания и размах, которые определяется множителем $e^{2x}$. Если движение идет вправо, то размах увеличивается, а при движении влево уменьшается. Вещественная и мнимая части сдвинуты относительно друг друга, поскольку множители $\cos(17x)$ и $\sin(17x)$ отличаются по фазе.
